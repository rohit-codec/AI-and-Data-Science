```python
import matplotlib.pyplot as plt 
import seaborn as sns 
```

# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

Let's go through examples of each!

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(x=df['sex'])

# x=df['sex'] → takes the 'sex' column for the X-axis
# countplot() → counts how many times each category occurs
# Here → counts the number of Male and Female customers
# Y-axis → shows the count/frequency
```

```python
sns.countplot(x=df['sex'], hue=df['smoker'])

# x=df['sex'] → shows Male and Female on the X-axis
# hue=df['smoker'] → further divides each sex by smoker category
# → smoker='Yes' and smoker='No' get separate bars/colors
# Y-axis → shows the count/frequency
# Useful for → comparing smoker/non-smoker counts across genders
```

```python
sns.barplot(x=df['sex'], y=df['total_bill'])

# x=df['sex'] → categories on the X-axis (Male, Female)
# y=df['total_bill'] → numerical values on the Y-axis
# barplot() → shows the average total_bill for each category
# Each bar → represents the mean (average) total_bill
```

```python
# You can change the estimator object to your own function, that converts a vector to a scalar
# Mean (default): np.mean
# Median: np.median
# Sum: np.sum
# Count: len (to count the number of observations)
# Standard Deviation: np.std
# Minimum: np.min
# Maximum: np.max
```

```python
import numpy as np

sns.barplot(x=df['sex'], y=df['tip'], estimator=np.sum)

# np.sum → calculates the sum instead of the default mean
# x=df['sex'] → categories: Male and Female
# y=df['tip'] → tip values
# estimator=np.sum → adds all tips for each category
# Each bar → shows the total tip given by Male/Female customers
```

# boxplot and violinplot

boxplots and violinplots are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.

```python
sns.boxplot(x='tip', y='day', data=df, palette='rainbow')

# x='tip' → numerical values (tip) are shown on the X-axis
# y='day' → categories (Thur, Fri, Sat, Sun) are shown on the Y-axis
# data=df → uses the df DataFrame
# boxplot() → shows the distribution of tips for each day
# Box → shows median, spread (IQR), and possible outliers
# palette='rainbow' → gives different colors to each category
```

```python
sns.violinplot(x='day', y='total_bill', data=df)

# x='day' → categories on the X-axis (Thur, Fri, Sat, Sun)
# y='total_bill' → numerical values on the Y-axis
# data=df → uses the df DataFrame
# violinplot() → shows the distribution of total_bill for each day
# Width/shape → shows where the data is more or less concentrated
# Wider part → more data values are concentrated there
# Narrower part → fewer data values are concentrated there
```

# stripplot and swarmplot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x='day', y='total_bill', data=df)

# x='day' → categories on the X-axis (Thur, Fri, Sat, Sun)
# y='total_bill' → numerical values on the Y-axis
# data=df → uses the df DataFrame
# stripplot() → shows individual data points for each category
# Each dot → represents one observation/customer
# Useful for → seeing the actual spread of data within each category
```

```python
sns.swarmplot(x='day', y='total_bill', data=df)

# x='day' → categories on the X-axis (Thur, Fri, Sat, Sun)
# y='total_bill' → numerical values on the Y-axis
# data=df → uses the df DataFrame
# swarmplot() → shows individual data points without overlapping
# Each dot → represents one observation/customer
# Unlike stripplot → points are adjusted so they don't overlap
```

```python
sns.violinplot(x='tip', y='day', data=df)

# violinplot() → shows the distribution of 'tip' for each day
# x='tip' → numerical values on X-axis
# y='day' → categories on Y-axis
# Shape/width → shows where tip values are concentrated


sns.swarmplot(x='tip', y='day', data=df)

# swarmplot() → adds individual data points on top
# Each dot → represents one customer's tip
# Dots are spread out to avoid overlapping
# Together → violinplot shows distribution + swarmplot shows actual observations
```
