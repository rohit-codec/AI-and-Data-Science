# Pandas Series

A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

## Creating Series
There are multiple ways to create a pandas Series:

```python
# %pip install pandas 
```

```python
import numpy as np 
import pandas as pd 
```

```python
my_list = [10,20,30]
pd.Series(my_list)
```

```python
labels = ['a', 'b', 'c']
pd.Series(my_list,index=labels) #series with custom labels 
```

```python
arr = np.array([10,20,30])
pd.Series(arr) #series from array
```

```python
d1 = {'a':10,'b':20,'c':30} #series from dictionary with index as kay and value as data point
d2 = {5:10,'c':20,9:30}

pd.Series(d1) , pd.Series(d2)
```
