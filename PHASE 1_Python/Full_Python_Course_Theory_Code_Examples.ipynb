````python
def withdraw(balance, amount):
    if amount < 0:
        raise ValueError("Amount cannot be negative")
    if amount > balance:
        raise ValueError("Insufficient balance")
    return balance - amount

try:
    print(withdraw(1000,1200))
except ValueError as e:
    print(e)
````

# 35. Decorators

A decorator wraps a function to add behavior without changing its main implementation.

````python
def logger(func):
    def wrapper(*args, **kwargs):
        print("Calling", func.__name__)
        result = func(*args, **kwargs)
        print("Finished")
        return result
    return wrapper

@logger
def add(a,b): return a+b
print(add(2,3))
````

# 44. Abstraction

`abc.ABC` and `@abstractmethod` let you define interfaces that concrete subclasses must implement.

````python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self): pass
class Rectangle(Shape):
    def __init__(self,w,h): self.w=w; self.h=h
    def area(self): return self.w*self.h
print(Rectangle(5,4).area())
````

# 67. DSA — Binary Search

Works on sorted data.

Time: O(log n).

Algorithm:
1. middle
2. compare
3. discard half
4. repeat

````python
def binary_search(a,target):
    l,r=0,len(a)-1
    while l<=r:
        m=(l+r)//2
        if a[m]==target:return m
        if a[m]<target:l=m+1
        else:r=m-1
    return -1

print(binary_search([1,3,5,7,9,11],7))
````

# 68. DSA — Sorting

Python's built-in sorting is stable and highly optimized.

For DSA interviews also implement:
- bubble sort
- selection sort
- insertion sort
- merge sort
- quicksort
- heap sort

````python
def insertion_sort(arr):
    a=arr.copy()
    for i in range(1,len(a)):
        key=a[i]; j=i-1
        while j>=0 and a[j]>key:
            a[j+1]=a[j]; j-=1
        a[j+1]=key
    return a

print(insertion_sort([5,2,4,1,3]))
````

# 70. DSA — Linked List

````python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

class LinkedList:
    def __init__(self):
        self.head=None
    def append(self,data):
        node=Node(data)
        if not self.head:
            self.head=node; return
        cur=self.head
        while cur.next:
            cur=cur.next
        cur.next=node
    def display(self):
        cur=self.head
        while cur:
            print(cur.data,end=" -> ")
            cur=cur.next
        print("None")

ll=LinkedList()
for x in [10,20,30]: ll.append(x)
ll.display()
````

# 71. DSA — Binary Tree

A binary tree node has at most two children.

Traversals:
- preorder: root-left-right
- inorder: left-root-right
- postorder: left-right-root
- level order: BFS

BST inorder traversal produces sorted order.

````python
class TreeNode:
    def __init__(self,val):
        self.val=val; self.left=None; self.right=None

def inorder(root):
    if root:
        inorder(root.left)
        print(root.val,end=" ")
        inorder(root.right)

root=TreeNode(10)
root.left=TreeNode(5)
root.right=TreeNode(15)
inorder(root)
````

# 72. DSA — Graph + BFS

Adjacency lists are common for sparse graphs.

BFS uses a queue and visits nodes level by level.

````python
from collections import defaultdict,deque
graph=defaultdict(list)
for u,v in [("A","B"),("A","C"),("B","D"),("C","D")]:
    graph[u].append(v); graph[v].append(u)

def bfs(start):
    seen={start}; q=deque([start])
    while q:
        u=q.popleft(); print(u,end=" ")
        for v in graph[u]:
            if v not in seen:
                seen.add(v); q.append(v)
bfs("A")
````

# 73. DSA — DFS

DFS can be implemented recursively or with an explicit stack.

````python
def dfs(graph,u,seen=None):
    if seen is None: seen=set()
    seen.add(u); print(u,end=" ")
    for v in graph[u]:
        if v not in seen: dfs(graph,v,seen)

dfs(graph,"A")
````

# 74. Backtracking

Backtracking explores choices and undoes a choice when returning.

Classic problems:
- subsets
- permutations
- combination sum
- N-Queens
- Sudoku

````python
def subsets(nums):
    ans=[]
    def bt(i,path):
        if i==len(nums):
            ans.append(path.copy()); return
        bt(i+1,path)
        path.append(nums[i])
        bt(i+1,path)
        path.pop()
    bt(0,[])
    return ans

print(subsets([1,2,3]))
````

# 83. Common Python Mistakes

### Mutable default arguments
Avoid:
```python
def f(items=[]):
    ...
```

Prefer:
```python
def f(items=None):
    if items is None:
        items=[]
```

### Other mistakes
- using `is` instead of `==` for value comparison
- modifying a list while iterating
- swallowing all exceptions with `except: pass`
- hard-coding secrets
- repeatedly using `pop(0)` instead of `deque`
- confusing shallow and deep copying

````python
def add_item(item, items=None):
    if items is None:
        items=[]
    items.append(item)
    return items

print(add_item("Python"))
print(add_item("C++"))
````
