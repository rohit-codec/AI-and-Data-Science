# how many country have Republic in their full name

```python
df.head(2)['country_long']
```

```python
# fun that counts the number of times the word 'republic' appears in the 'country_long' column
count = 0
def counting(txt):
    global count  # since count variable is not accessible outside the function so we need to declare it as global variable so that it can be accessed outside the function.
    if 'republic' in txt.lower():
        count+= 1
    return txt
```

```python
# This will apply the counting function to each value in the 'country_long' column and update the column with the returned values
# note this will not change the original 'country_long' column, it will just apply the function and return the modified values. 
# we when we run this code it also print text for each row of country_long column and also count the number of times the word 'republic' appears in the 'country_long' column
#so we need not to print the text for each row of country_long column so we assigned the returned values to the same column 'country_long' so that it will not print the text for each row of country_long column. 
df['country_long'] = df['country_long'].apply(counting)
```

```python
# but count vb works well after the apply function is executed because the apply function will call the counting function for each row of the 'country_long' column and update the count variable accordingly.

# so we can print the count variable after the apply function is executed to get the number of times the word 'republic' appears in the 'country_long' column.
print(count)
```
