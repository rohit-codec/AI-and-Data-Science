```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

We use distribution plots to visualize the distribution of quantitative data.


Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

```python
df = sns.load_dataset('tips')  # seaborn lib ke pass khud data hai bs name se load kr lo
df
```

```python
df['size'].unique()
```

```python
df['total_bill']
```

```python
sns.histplot(df['total_bill'], bins=20)

# sns.histplot() → creates a histogram
# df['total_bill'] → numerical data to plot
# bins=20 → divides the data range into 20 intervals (bars)
# Histogram → shows the distribution/frequency of total_bill
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True) 
# kde=True → adds a KDE (Kernel Density Estimate) curve
# KDE → shows a smooth curve representing the distribution of the data
# Histogram → shows frequency using bars
# KDE → shows the overall shape of the distribution smoothly

plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)

```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'scatter')

# x='total_bill' → total_bill is shown on the X-axis
# y='tip' → tip is shown on the Y-axis
# data=df → use the df DataFrame
# kind='scatter' → creates a scatter plot
# Jointplot → shows the relationship/distribution between two variables
# Each dot → represents one observation/row
```

```python
sns.jointplot(x='total_bill', y='tip', data=df, kind='reg')

# x='total_bill' → total_bill is on the X-axis
# y='tip' → tip is on the Y-axis
# data=df → uses the df DataFrame
# kind='reg' → creates a scatter plot + regression line
# Regression line → shows the overall relationship/trend between total_bill and tip
# Side plots → show the distribution of each variable
```

```python
sns.pairplot(df)

# pairplot() → creates plots for all numerical columns in the DataFrame
# Diagonal → shows the distribution of each variable (histogram)
# Off-diagonal → shows relationship between every pair of variables (scatter plot)
# Useful for → quickly understanding relationships and distributions
```

```python
sns.pairplot(df, hue='size', palette='rainbow')

# df → DataFrame to visualize
# hue='size' → separates/color-codes the data according to the 'size' column
# palette='rainbow' → uses the rainbow color palette for different size categories
# pairplot() → shows relationships between all numerical columns
```

```python
sns.rugplot(df['tip'])

# df['tip'] → uses the tip column
# rugplot() → shows the distribution of data using small marks on the axis
# Each small line → represents one data point
# Useful for → seeing where values are concentrated
```
