*Deep and Shallow copy*

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice *10 # copy og mai change nhi hoga in 2d make shallow copy
slice , a
```

```python
a[4:6] *10 , a  # copy og mai change nhi hoga
```

```python
a
b = a #make reference
b[0] = 99 
b,a
```

```python
a
b = a.copy() #make copy
b[0] = 1 
b,a
```

**Advance Array Manipulation**

*Stacking Array* onlu use vector

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8]) 
a , b
```

```python
c=np.vstack((a,b)) #vertical stack same dimensional ka array hona chaihye
c
```

```python
d=np.array((c,c))
d
```

```python
np.hstack((a,b))  #horizontal stack
```

```python
np.hstack((c,c)) 
```

```python
np.column_stack((a,b))  #column stack
```
