# I) make a new column for episode count

**we extract episode number from title of each column**



```python
# first we will create a function to extract the number of episodes from the 'Episodes' column. The 'Episodes' column contains the number of episodes in the format 'x (y)', where x is the number of episodes and y is the number of seasons. We will extract the number of episodes from this string.
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True: 
            data = data + i 
        if i == '(':
            check = True 
```

```python
df["Title"].apply(extract_episodes)
```

```python
#add extracted episodes to the dataframe as a new column called 'Episodes'
df['Episodes']= df["Title"].apply(extract_episodes)
df
```

**we  receive as x eps , first we convert  into x , since this x is a str type we convert it into integer type**

```python
# Now we will remove the " eps" from the 'Episodes' column. The 'Episodes' column contains the number of episodes in the format 'x eps', where x is the number of episodes. We will remove the " eps" from this string.
df['Episodes']= df['Episodes'].str.replace(" eps","") #replace the " eps" with an empty string in the 'Episodes' column
df 
```

# II) make a new column for time stamp

```python
df.loc[3]['Title'] # displays the value in the 'Title' column of the 4th row of the dataframe. You can also specify the index of the row you want to display by passing an integer as an argument to the loc[] function. For example, df.loc[10]['Title'] will display the value in the 'Title' column of the 11th row of the dataframe.
```

```python
#fun that extracts the time stamp from the 'Title' column. The 'Title' column contains the time stamp in the format '(x)y,', where x is the episode number of the anime and y is the time stamp. We will extract the time stamp from this string.
def extraction_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')': # #since the time stamp is always after the ')'
            for j in range(i+1,i+20):  # size of the time stamp is equal to 18 characters
                data += txt[j]

            return data       
```

```python
df['Total Time'] = df['Title'].apply(extraction_time) #add extracted time stamp to the dataframe as a new column called 'Total Time'
```

```python
df.head(2)
```

```python
#calculate the total number of months between the start and end dates in the 'Total Time' column. The 'Total Time' column contains the time period in the format 'MMM YYYY - MMM YYYY', where MMM is the month and YYYY is the year. We will calculate the total number of months between the start and end dates in this string.
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1  # +1 to include the starting month
    except:
        return None

df['Months'] = df['Total Time'].apply(calculate_total_months)
df.head(3)
```
