# Types of Data for Plotting

## 1. Qualitative Data
- Data that describes **categories or qualities**.
- Usually represented using **words/text** rather than numbers.
- Examples:
  - Gender → Male, Female
  - Color → Red, Blue, Green
  - City → Delhi, Mumbai, Chennai

### Common Plots:
- Bar Plot
- Count Plot
- Pie Chart


## 2. Quantitative Data
- Data that represents **numerical values**.
- It can be measured or counted.
- Examples:
  - Age → 20, 25, 30
  - Salary → 30000, 50000
  - Height → 170.5, 175.2

### Common Plots:
- Histogram
- Box Plot
- Scatter Plot
- Line Plot


## 3. Categorical Data
- Data consisting of a **limited number of distinct categories**.
- Categories may be text or labels.
- Examples:
  - Department → IT, HR, Sales
  - Size → Small, Medium, Large
  - Rating → Good, Average, Poor

### Common Plots:
- Bar Plot
- Count Plot
- Pie Chart


## 4. Distribution Data
- Shows **how numerical values are spread or distributed**.
- It helps us understand:
  - Center of the data
  - Spread of the data
  - Shape of the data
  - Outliers
  - Skewness

- Example:
  - Marks → 45, 50, 52, 60, 65, 70, 85, 90

### Common Plots:
- Histogram → Shows frequency distribution
- KDE Plot → Shows smooth distribution
- Box Plot → Shows spread, median, and outliers
- Violin Plot → Shows distribution + spread


## Quick Summary

| Data Type | Meaning | Common Plot |
|----------|---------|-------------|
| Qualitative | Descriptive/category-based data | Bar, Pie |
| Quantitative | Numerical data | Histogram, Scatter, Box |
| Categorical | Data divided into categories | Bar, Count |
| Distribution | How numerical data is spread | Histogram, KDE, Box, Violin |
