```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
tips = sns.load_dataset("tips")
```

# **Regression Plot**

**Regression Plot → shows the relationship between two numerical variables.**

**It contains → scatter points + a regression line.**

**Regression line → shows the overall trend/relationship between X and Y.**

**sns.regplot() → is used to create a regression plot in Seaborn.**

**Positive slope → positive relationship.**

**Negative slope → negative relationship.**

# lmplot()
*regression plot is used in machine learning so we will not fully discuss what, how and where it is used
but we will see how it is created and what are the values we can put inside it*

```python
sns.lmplot(x='total_bill', y='tip', data=tips)

# x='total_bill' → numerical variable on the X-axis
# y='tip' → numerical variable on the Y-axis
# data=tips → uses the tips DataFrame
# lmplot() → creates a scatter plot with a regression line
# Regression line → shows the overall relationship/trend between total_bill and tip
# lmplot() → useful for creating regression plots with additional grouping options
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex')

# x='total_bill' → numerical variable on the X-axis
# y='tip' → numerical variable on the Y-axis
# data=tips → uses the tips DataFrame
# hue='sex' → separates the data based on sex
# → different colors and separate regression lines for Male and Female
# lmplot() → creates scatter points + regression line
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette='coolwarm')
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette='coolwarm',markers=['o','v'],
           scatter_kws={'s':50})
```
