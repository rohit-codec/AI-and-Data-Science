```python
import seaborn as sns
import pandas as pd
import cufflinks as cf

cf.go_offline()
from plotly.offline import iplot


tips = sns.load_dataset('tips')
tips
```

```python
import sys
import numpy as np
import pandas as pd
import plotly
import cufflinks

print("Python:", sys.version)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Plotly:", plotly.__version__)
print("Cufflinks:", cufflinks.__version__)
```

```python
import plotly.express as px

fig = px.histogram(tips, x='total_bill')
fig.show()
```

```python
tips['total_bill'].iplot()
```

```python
tips.groupby('day')['tip'].mean().iplot(kind = 'bar')
```

```python
tips.iplot(kind='scatter',x ='total_bill',y = 'tip',mode = 'markers')
```

```python
tips[['total_bill','day']].iplot(kind = 'box',x = 'total_bill',y = 'day')
```
