```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

# Matplotlib Basics

```python
x = np.linspace(0,5,11) #make an array of 11 values from 0 to 5 with equal spacing
x
```

```python
y = x **3 # make an array of 11 values from 0 to 5 with equal spacing and then cube each value in the array
y
```

**Line graph**

```python
plt.plot(x,y) #plot the values of x and y on a graph then  it makes a line graph with x values on the x-axis and y values on the y-axis
plt.title("y=x^3") #add a title to the graph
plt.xlabel("X-axis") #add a label to the x-axis
plt.ylabel("Y-axis") #add a label to the y-axis
```

**Subplots in Matplotlib**

```python
#creation of subplots and subplotting of graphs in the subplots note that the order of the subplots is from left to right and then from top to bottom
# note subplot number hmesha dena hota hai


plt.subplot(2,2,1) #first select subplot
plt.plot(x,y)       #then plot graph on that subplot

plt.subplot(2,2,4)   
plt.plot(x,y)

plt.subplot(2,2,3)


plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

# Object Oriented way

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches
dpi is the dots-per-inch (pixel per inch).

```python
fig = plt.figure(figsize=(20,8),dpi = 100) # change canvas size and dpi -clearity

fig=plt.figure() # it creates empty canvas

# ek hi canvs pr kai saare
# values btw 0 to 1

axis1 = fig.add_axes([0.1,0.1,0.6,0.8])  #[ Left position of the axes, Bottom position of the axes, Width of the axes, Height of the axes]
axis1.plot(x,y)


axis2 = fig.add_axes([0.3,0.4, 0.3,  0.3])
#             │    │    │    │
#             │    │    │    └── 30% of figure height
#             │    │    └─────── 30% of figure width
#             │    └──────────── 40% from bottom
#             └───────────────── 30% from left
axis2.plot(x,y)
```
